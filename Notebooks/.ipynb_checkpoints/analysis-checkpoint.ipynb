```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import chi2_contingency

from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv('../Data/cw_22_23_24.csv')
df.head(5)
```

```python
missing_data = df.isnull().sum()
print(missing_data)
```

```python
missing_di_data = df['disease'].isnull().sum()
print(missing_di_data)
```

```python
total_diseases = df['disease'].count()
total_diseases
```

```python
perc_of_missing_data = missing_di_data/total_diseases
perc_of_missing_data
```

```python
df['disease'] = df['disease'].fillna('Missing')

contingency = pd.crosstab(df['disease'], df['status'])

chi2, p, dof, expected = chi2_contingency(contingency)

print("Chi-square statistic:", chi2)
print("Degrees of freedom:", dof)
print("p-value:", p)

```

```python
count_dupe = df.duplicated().sum()
count_dupe
```

```python


# 1️⃣ Encode the target
df['status_num'] = LabelEncoder().fit_transform(df['status'])

# 2️⃣ One-hot encode predictors
X = pd.get_dummies(
    df.drop(columns=['status']),  # drop original target
    drop_first=True,
    dummy_na=True
)

# 3️⃣ Correlation of each feature with the target
target_corr = X.corrwith(df['status_num']).sort_values(key=abs, ascending=False)

# 4️⃣ Display top correlations
print(target_corr)
```
